```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
with open(('../pytorch-for-deep-learning/Data/shakespeare.txt'), 'r', encoding='utf-8') as f:
    text = f.read()
print('Length of dataset in characters:', len(text))
```

```python
print('First 1000 characters:\n', text[:1000])
```

```python
all_characters = set(text)
print('All characters:', all_characters)
```

```python
decoder = dict(enumerate(all_characters))
```

```python
decoder
```

```python
encoder = {v: k for k, v in decoder.items()}
encoder
```

```python
encoded_text = np.array([encoder[c] for c in text], dtype=np.int32)
encoded_text[:1000]
```

```python
decoder[9] # space
```

```python
def one_hot_encoder(encoded_text, num_uni_chars):
    one_hot = np.zeros((len(encoded_text), num_uni_chars), dtype=np.float32)
    one_hot[np.arange(len(encoded_text)), encoded_text] = 1
    return one_hot
num_uni_chars = len(all_characters)
one_hot = one_hot_encoder(encoded_text, num_uni_chars)
print('One-hot shape:', one_hot.shape)
```
