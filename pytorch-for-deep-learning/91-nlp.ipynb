```python
def one_hot_encoder(encoded_text, num_uni_chars):
    one_hot = np.zeros((len(encoded_text), num_uni_chars), dtype=np.float32)
    one_hot[np.arange(len(encoded_text)), encoded_text] = 1
    return one_hot
num_uni_chars = len(all_characters)
one_hot = one_hot_encoder(encoded_text, num_uni_chars)
print('One-hot shape:', one_hot.shape)
```

```python
def generate_batches(encoded_text, samp_per_batch=10, seq_len=50):
    # X : encoded text of length seq_len
    # Y : encoded text shifted by one

    char_per_batch = samp_per_batch * seq_len
    num_batches = len(encoded_text) // char_per_batch
    encoded_text = encoded_text[:num_batches * char_per_batch]
    encoded_text = encoded_text.reshape((samp_per_batch, -1))
    for n in range(0, encoded_text.shape[1], seq_len):
        x = encoded_text[:, n:n + seq_len]
        y = encoded_text[:, n + 1:n + seq_len + 1]
        yield x, y
# Check the function
samp_per_batch = 2
seq_len = 10
x, y = next(generate_batches(encoded_text, samp_per_batch, seq_len))
print('X shape:', x.shape)
print('Y shape:', y.shape)
# Check the first batch
print('X:', x)
print('Y:', y)
```
